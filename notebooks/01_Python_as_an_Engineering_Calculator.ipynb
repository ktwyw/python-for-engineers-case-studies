## 1. Python does arithmetic

The symbols `+`, `-`, `*` and `/` work as on a calculator. `**` raises to a power. Python follows the usual order of operations: powers first, then multiplication and division, then addition and subtraction. Parentheses override the order.

```python
# Addition, subtraction, multiplication and division, as on a calculator.
print(7 + 2, 7 - 2, 7 * 2, 7 / 2)
# ** is "to the power of": 2 cubed is 8.
print(2 ** 3)
# Multiplication happens before addition: this is 2 + 12 = 14, not 20.
print(2 + 3 * 4)
# Parentheses change the order: (2 + 3) * 4 = 20.
print((2 + 3) * 4)
```

Two more operators are very useful for counting things. `//` divides and throws away the fractional part (**floor division**), and `%` gives the **remainder**.

A pipe run of 23 m is built from standard 6 m lengths. How many full lengths fit, and how much is left over?

```python
# Total length of the pipe run in metres.
run_length = 23
# Length of one standard pipe section in metres.
section_length = 6
# Floor division: how many whole sections fit into the run.
full_sections = run_length // section_length
# Remainder: the metres still to cover after the full sections.
leftover = run_length % section_length
# Print both results.
print("full sections:", full_sections, "| leftover length:", leftover, "m")
# The leftover needs one more (cut) section, so 3 + 1 = 4 sections are ordered.
print("sections to order:", full_sections + 1)
```

## 2. Variables: names for quantities

Typing raw numbers again and again invites mistakes. A **variable** is a name that refers to a value. The line `g = 9.81` does not mean "g equals 9.81" as in algebra. It means *store 9.81 and call it `g`*. Good names say what the quantity is, and a comment gives its unit.

Our tank: inner diameter 2.0 m, water depth 3.5 m, water at about 20 °C.

```python
# Density of water at about 20 °C, in kg/m³.
rho_water = 998.0
# Gravitational acceleration, in m/s².
g = 9.81
# Inner diameter of the tank, in m.
tank_diameter = 2.0
# Depth of water in the tank, in m.
water_depth = 3.5
# Atmospheric pressure at sea level, in Pa.
p_atm = 101325.0
# Show the stored values; print separates its arguments with spaces.
print(rho_water, g, tank_diameter, water_depth, p_atm)
```

## 3. Hydrostatic pressure on the tank floor

The weight of the water column above the floor creates a pressure

$$p_\text{gauge} = \rho g h ,$$

where $\rho$ is the density, $g$ the gravitational acceleration and $h$ the depth. This is the **gauge** pressure: the pressure above atmospheric. A pressure gauge reads this value. The **absolute** pressure adds the atmosphere on top: $p_\text{abs} = p_\text{atm} + \rho g h$.

> **Predict first:** is the gauge pressure closer to 3 kPa, 30 kPa or 300 kPa?

```python
# Gauge pressure on the floor: density × gravity × depth, in Pa.
p_gauge = rho_water * g * water_depth
# Absolute pressure: add the atmospheric pressure.
p_abs = p_atm + p_gauge
# Print both values in pascals.
print("gauge pressure    =", p_gauge, "Pa")
# The absolute pressure is always larger than the gauge pressure.
print("absolute pressure =", p_abs, "Pa")
```

The raw output prints every digit Python has. Engineers report a sensible number of digits, with units. An **f-string** (a string with `f` before the opening quote) inserts values into text. Inside the braces, `:.1f` means "one digit after the decimal point", `:,.0f` adds thousands separators, and `:.3e` gives scientific notation.

```python
# Convert Pa to kPa by dividing by 1000, and keep one decimal.
print(f"gauge pressure    = {p_gauge / 1000:.1f} kPa")
# Thousands separators make large numbers easier to read.
print(f"absolute pressure = {p_abs:,.0f} Pa")
# Scientific notation: 1.356e+05 means 1.356 × 10⁵.
print(f"absolute pressure = {p_abs:.3e} Pa")
```

## 4. Types: whole numbers, decimal numbers, text and true/false

Every value in Python has a **type**:

- `int`: whole numbers, such as `6`;
- `float`: numbers with a decimal point, such as `9.81`;
- `str`: text, written in quotes, such as `"Tank T-101"`;
- `bool`: the two values `True` and `False`.

The function `type()` reports a value's type. The type decides what operations make sense.

```python
# A whole number: an int.
print(type(section_length))
# A number with a decimal point: a float.
print(type(g))
# Text in quotes: a str.
tank_name = "Tank T-101"
# Show the type of the text.
print(type(tank_name))
# A comparison produces a bool: is the depth more than 3 m?
print(water_depth > 3, type(water_depth > 3))
```

Two details trip up beginners:

- `/` **always** gives a float, even when the division comes out even. `//` gives an int when both numbers are ints.
- Text that looks like a number is still text. A value read from a file or a form arrives as a string, and `float()` or `int()` converts it. `int()` drops the fractional part; it does not round.

```python
# / always gives a float: 6 / 2 is 3.0, not 3.
print(6 / 2, type(6 / 2))
# // with two ints gives an int.
print(7 // 2, type(7 // 2))
# A depth that arrived as text, for example from a data file.
depth_text = "3.5"
# Convert the text to a float before calculating with it.
depth_from_text = float(depth_text)
# Now arithmetic works: twice the depth.
print(depth_from_text * 2)
# int() truncates towards zero: 3.9 becomes 3, not 4.
print(int(3.9), round(3.9))
```

## 5. The `math` module: volume and mass of the water

Python's built-in tools stop at basic arithmetic. The `math` module adds $\pi$, square roots, logarithms, trigonometry and more. You make it available with `import math`, then write `math.pi`, `math.sqrt(x)` and so on.

The water volume of a cylinder is $V = \pi r^2 h$, and its mass is $m = \rho V$.

```python
# Load the standard math module.
import math
# The radius is half the diameter.
tank_radius = tank_diameter / 2
# Volume of the water: π r² h, in m³.
volume = math.pi * tank_radius ** 2 * water_depth
# Mass of the water: density × volume, in kg.
mass = rho_water * volume
# 1 m³ is 1000 litres.
litres = volume * 1000
# Report the volume in m³ and litres.
print(f"volume = {volume:.2f} m³ = {litres:,.0f} L")
# Report the mass in kg and tonnes.
print(f"mass   = {mass:,.0f} kg = {mass / 1000:.1f} t")
```

## 6. How long does the tank take to drain?

A hole of diameter $d$ is opened at the bottom. **Torricelli's law** says the water leaves at speed $\sqrt{2gh}$, the speed of a body falling from height $h$. A real sharp-edged hole delivers only part of that ideal flow; a **discharge coefficient** of about $C_d = 0.61$ accounts for the narrowing jet. As the level falls, the outflow slows. Solving the resulting equation gives the time to empty the tank completely:

$$t_\text{drain} = \frac{A_\text{tank}}{C_d\,A_\text{hole}}\sqrt{\frac{2h}{g}} .$$

You do not need to derive this. Here it is an exercise in typing a formula correctly. (Notebook 02 checks it by simulating the draining step by step.)

```python
# Diameter of the drain hole, in m (5 cm).
hole_diameter = 0.05
# Discharge coefficient of a sharp-edged hole (dimensionless).
c_d = 0.61
# Cross-sectional area of the tank, π r², in m².
area_tank = math.pi * tank_radius ** 2
# Area of the hole, π (d/2)², in m².
area_hole = math.pi * (hole_diameter / 2) ** 2
# Torricelli's drain-time formula, in seconds.
t_drain = area_tank / (c_d * area_hole) * math.sqrt(2 * water_depth / g)
# Report the time in seconds and in minutes.
print(f"drain time = {t_drain:.0f} s = {t_drain / 60:.1f} min")
```

A time of 2216 s is easier to understand as minutes and seconds. Floor division and remainder from Section 1 do exactly this.

```python
# Round the time to whole seconds and make it an int.
total_seconds = round(t_drain)
# Whole minutes: floor division by 60.
minutes = total_seconds // 60
# Seconds left over after the whole minutes.
seconds = total_seconds % 60
# Print as minutes and seconds.
print(f"drain time = {minutes} min {seconds} s")
# Check: putting the pieces back together gives the total.
assert minutes * 60 + seconds == total_seconds
```

The last line is a **check**. `assert` followed by a condition does nothing if the condition is `True`, and stops the program with an error if it is `False`. Every notebook in this series uses checks like this. A program is not finished when it runs; it is finished when you have reason to believe its answer.

## 7. Unit conversions

Pressure gauges are labelled in kPa, bar or psi, depending on the country and industry. Storing each conversion factor once, in a clearly named variable, avoids retyping (and mistyping) it.

```python
# Pascals in one bar.
PA_PER_BAR = 100000.0
# Pascals in one pound-force per square inch (psi).
PA_PER_PSI = 6894.757
# Gauge pressure in bar.
p_gauge_bar = p_gauge / PA_PER_BAR
# Gauge pressure in psi.
p_gauge_psi = p_gauge / PA_PER_PSI
# Report all three units together.
print(f"gauge pressure = {p_gauge / 1000:.1f} kPa = {p_gauge_bar:.3f} bar = {p_gauge_psi:.2f} psi")
# Check by converting back: bar → Pa must return the original value.
assert math.isclose(p_gauge_bar * PA_PER_BAR, p_gauge)
```

Capital letters in `PA_PER_BAR` are a Python convention for **constants**: values that should never change while the program runs.

## 8. Decimal numbers are stored approximately

Computers store floats in binary, and most decimal fractions, such as 0.1, have no exact binary form. The tiny errors are usually harmless, but they mean you should **never test two floats for exact equality**. Use `math.isclose`, which allows a small relative difference.

```python
# 0.1 + 0.2 is not exactly 0.3 in binary arithmetic.
print(0.1 + 0.2)
# So an exact comparison gives False.
print(0.1 + 0.2 == 0.3)
# math.isclose allows a tiny relative difference and gives True.
print(math.isclose(0.1 + 0.2, 0.3))
# The same applies to engineering results: compare with a tolerance.
print(math.isclose(p_gauge, 34267.0, rel_tol=1e-4))
```

## 9. Comparisons give `True` or `False`

The tank's floor is rated for a gauge pressure of 40 kPa. A comparison such as `p_gauge < limit` produces a `bool`. Notebook 02 uses such results to make decisions with `if`.

```python
# Rated gauge pressure of the floor, in Pa.
p_limit = 40000.0
# Is the floor within its rating? This is a bool.
within_rating = p_gauge < p_limit
# Print the answer.
print("floor within rating:", within_rating)
# How full could the tank be at the limit? Rearranged: h = p / (ρ g).
max_depth = p_limit / (rho_water * g)
# Report the maximum depth.
print(f"maximum water depth at the rating = {max_depth:.2f} m")
```

## 10. Reading error messages

Mistakes are normal. Python stops and reports an **error** with its type and a short message. Learning to read that message is half of programming. Two common ones:

- `NameError`: a name that was never defined, often a typo, such as `water_dpth`.
- `TypeError`: an operation that makes no sense for the types involved, such as adding text to a number.

The cell below makes both mistakes on purpose. The `try`/`except` lines are a safety net that catches the error, so the notebook keeps running; you will meet them properly in Notebook 08.

```python
# Try a line that contains a typo in a variable name ...
try:
    # ... "water_dpth" was never defined.
    print(water_dpth)
# Catch the NameError and print Python's own message.
except NameError as error:
    # Show the error type and message.
    print("NameError:", error)
# Try adding a string to a number ...
try:
    # ... text "3.5" plus the number 1 is not defined.
    print(depth_text + 1)
# Catch the TypeError and print Python's message.
except TypeError as error:
    # Show the error type and message.
    print("TypeError:", error)
```
